```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/IBM_TCC.csv")

print("Shape:", df.shape)
df.head()
```

```python
df.info()
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nBlank/whitespace values:")
print(
    df.apply(
        lambda col: col.astype(str).str.strip().eq('').sum()
    )
)
```

```python
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"].astype(str).str.strip(),
    errors="coerce"
)

print("Missing TotalCharges after conversion:", df["TotalCharges"].isnull().sum())
```

```python
df["TotalCharges"] = df["TotalCharges"].fillna(0)
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum())

print("\nData types after cleaning:")
print(df.dtypes)
```

```python
duplicate_ids = df["customerID"].duplicated().sum()

print("Duplicate customer IDs:", duplicate_ids)
```

```python
phone_violations = (
    (df["PhoneService"] == "No") &
    (df["MultipleLines"] != "No phone service")
).sum()

internet_columns = [
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

internet_violations = sum(
    (
        (df["InternetService"] == "No") &
        (df[column] != "No internet service")
    ).sum()
    for column in internet_columns
)

print("Phone service violations:", phone_violations)
print("Internet service violations:", internet_violations)
```

```python
print("Final shape:", df.shape)

print("\nFinal data types:")
print(df.dtypes)

print("\nFinal missing values:")
print(df.isnull().sum().sum())
```

```python
output_path = "../data/processed/telco_customer_churn_cleaned.csv"

df.to_csv(output_path, index=False)

print(f"Cleaned dataset saved to: {output_path}")
```
